```python
import pandas as pd
import numpy as np

# Dataset karyawan (sengaja mengandung masalah kualitas)
data = {
    'usia'      : [25, 32, np.nan, 45, 28, 51, 38, np.nan, 29, 41],
    'pendapatan': [4_000_000, 7_500_000, 5_200_000, 12_000_000, 4_800_000,
                   15_000_000, 8_100_000, 6_300_000, np.nan, 9_900_000],
    'pendidikan': ['SMA','S1','SMA','S2','SMA','S2','S1','S1','SMA','S2'],  # ordinal
    'kota'      : ['Bandung','Jakarta','Bandung','Surabaya','Jakarta',
                   'Surabaya','Jakarta','Bandung','Bandung','Jakarta'],    # nominal
    'membeli'   : ['Tidak','Ya','Tidak','Ya','Tidak','Ya','Ya','Tidak','Tidak','Ya']  # target
}
df = pd.DataFrame(data)
df
```

# LATIHAN MANDIRI

### 1. Ganti imputasi usia menjadi mean; bandingkan hasilnya.

```python
df_median = df.copy()
df_median['usia'] = df_median['usia'].fillna(df_median['usia'].median())
df_median['pendapatan'] = df_median['pendapatan'].fillna(df_median['pendapatan'].median())

print("median usia:", df_median['usia'].median())
print("median pendapatan:", df_median['pendapatan'].median())
print(display(df_median))
```

```python
df['usia'] = df['usia'].fillna(df['usia'].mean())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].mean())

print("median usia:", df['usia'].mean())
print("median pendapatan:", df['pendapatan'].mean())
print(display(df))
```

```python
print("Median:", df_median['usia'].median())
print("Mean:", df['usia'].mean())
```

### 3. Tambahkan kolom kategorikal baru dan terapkan one-hot encoding.

```python
X = df.drop(columns=['membeli'])
y = df['membeli']
X['status_karyawan'] = ['Tetap','Kontrak','Tetap','Tetap','Kontrak','Tetap','Kontrak','Tetap','Kontrak','Tetap']

#Encoding kategorikal (ordinal & one-hot).
X['status_karyawan'] = X['status_karyawan'].map({'Tetap':1, 'Kontrak':0})
X['pendidikan'] = X['pendidikan'].map({'SMA':0,'S1':1,'S2':2}) # ordinal
X = pd.get_dummies(X, columns=['kota'], dtype=int) # nominal
y = y.map({'Tidak':0,'Ya':1})
display(X)
display(y)
```

### 2. Gunakan MinMaxScaler (0–1) sebagai ganti StandardScaler; tampilkan describe

```python
#Membagi data (SEBELUM scaling).

from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(
X, y, test_size=0.3, random_state=42)

#Penskalaan (fit pada latih, transform pada latih & uji).

from sklearn.preprocessing import MinMaxScaler
num = ['usia','pendapatan','pendidikan']
sc = MinMaxScaler()
X_train[num] = sc.fit_transform(X_train[num])
X_test[num] = sc.transform(X_test[num])  # cegah data leakage

print("X_train setelah MinMaxScaler")
print(display(X_train))

print("X_test setelah MinMaxScaler")
print(display(X_test))

print("X_train describe")
display(X_train[num].describe())
```

### 4. Jelaskan mengapa fit_transform pada latih tetapi hanya transform pada uji.

Jawab:

it_transform digunakan pada data latih karena data latih digunakan untuk mencari nilai yang dibutuhkan dalam proses scaling. Sedangkan pada data uji cukup menggunakan transform karena data uji hanya mengikuti hasil scaling dari data latih. Hal ini dilakukan supaya tidak terjadi data leakage, yaitu data uji ikut memengaruhi proses pembelajaran.

###1. Mengapa urutan 'split dulu, baru scaling' penting?
####Jawab:
split dulu baru scaling penting supaya data uji tidak ikut memengaruhi proses scaling. Scaling cukup dipelajari dari data latih, sehingga hasil pengujian lebih objektif dan tidak terjadi data leakage.
###2. Kapan label encoding dan kapan one-hot encoding?
####Jawab:
label encoding digunakan untuk data kategorikal yang memiliki urutan, seperti SMA < S1 < S2. Sedangkan one-hot encoding digunakan untuk data kategorikal yang tidak memiliki urutan, seperti Bandung, Jakarta, dan Surabaya.
###3. Apa perbedaan normalisasi dan standardisasi?
####Jawab:
Normalisasi mengubah nilai data ke rentang tertentu, biasanya 0 sampai 1. Sedangkan standardisasi mengubah data berdasarkan rata-rata dan standar deviasi, sehingga data memiliki rata-rata sekitar 0 dan standar deviasi 1.
